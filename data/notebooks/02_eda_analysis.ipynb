```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv("C:/Users/eluma/Desktop/restaurant-ai-project/data/cleaned_food_demand.csv")
```

```python
df.head()
```

```python
weekly_orders = df.groupby('week')['num_orders'].sum()

plt.figure(figsize=(12,6))
plt.plot(weekly_orders.index, weekly_orders.values)
plt.title("Weekly Food Orders Trend")
plt.xlabel("Week")
plt.ylabel("Number of Orders")
plt.show()
```

The weekly food orders trend shows fluctuating customer demand patterns across diffrent weeks.

```python
top_meals = df.groupby('meal_id')['num_orders'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
top_meals.plot(kind='bar')

plt.title("Top Selling Meals")
plt.xlabel("Meal ID")
plt.ylabel("Total Orders")

plt.show()
```

Certain meal categories consistently recieve higher customer demand, identifying top-selling meals helps restaurants priortize inventory planning and optimize menu offerings.
