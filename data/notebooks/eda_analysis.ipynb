```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('cleaned_restaurant_sales.csv')

print(df.head())
```

```python
daily_sales = df.groupby('Date')['Quantity_Sold'].sum()

plt.figure(figsize=(12,5))

plt.plot(daily_sales)

plt.title('Daily Sales Trend')

plt.xlabel('Date')

plt.ylabel('Quantity Sold')

plt.show()
```

```python
top_items = df.groupby('Food_Item')['Quantity_Sold'].sum()

top_items = top_items.sort_values(ascending=False)

plt.figure(figsize=(10,5))

top_items.plot(kind='bar')

plt.title('Top Selling Food Items')

plt.xlabel('Food Item')

plt.ylabel('Total Quantity Sold')

plt.show()
```

```python
monthly_sales = df.groupby('Month')['Quantity_Sold'].sum()

plt.figure(figsize=(10,5))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Analysis')

plt.xlabel('Month')

plt.ylabel('Quantity Sold')

plt.show()
```

```python
weekend_sales = df.groupby('Weekend')['Quantity_Sold'].sum()

plt.figure(figsize=(6,5))

weekend_sales.plot(kind='bar')

plt.title('Weekend vs Weekday Sales')

plt.xlabel('Weekend')

plt.ylabel('Quantity Sold')

plt.show()
```

```python
weather_sales = df.groupby('Weather')['Quantity_Sold'].sum()

plt.figure(figsize=(8,5))

weather_sales.plot(kind='bar')

plt.title('Weather Impact on Sales')

plt.xlabel('Weather')

plt.ylabel('Quantity Sold')

plt.show()
```

# EDA Summary

The analysis shows that food demand varies based on:
- Time
- Weather conditions
- Weekend patterns
- Customer traffic

These insights will help train machine learning models for demand forecasting and inventory optimization.
