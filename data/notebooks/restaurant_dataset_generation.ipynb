```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

```python
num_rows = 5000
print("Dataset Rows:", num_rows)
```

```python
food_items = [
    'Burger',
    'Pizza',
    'Pasta',
    'Biryani',
    'Sandwich',
    'Fried Rice',
    'Noodles',
    'Coffee',
    'Tea',
    'Ice Cream',
    'Cake',
    'Salad',
    'Wrap',
    'Tacos',
    'Juice'
]

print(food_items)
```

```python
weather_conditions = [
    'Sunny',
    'Rainy',
    'Cloudy',
    'Hot'
]

print(weather_conditions)
```

```python
start_date = datetime(2023, 1, 1)

dates = [
    start_date + timedelta(days=random.randint(0, 730))
    for _ in range(num_rows)
]

print(dates[:5])
```

```python
restaurant_data = {

    'Order_ID': range(1, num_rows + 1),

    'Date': dates,

    'Food_Item': np.random.choice(food_items, num_rows),

    'Quantity_Sold': np.random.randint(1, 20, num_rows),

    'Price': np.round(
        np.random.uniform(50, 500, num_rows), 2
    ),

    'Weather': np.random.choice(
        weather_conditions,
        num_rows
    ),

    'Weekend': np.random.choice(
        [0, 1],
        num_rows
    ),

    'Festival': np.random.choice(
        [0, 1],
        num_rows
    ),

    'Customer_Count': np.random.randint(
        20,
        300,
        num_rows
    ),

    'Inventory_Level': np.random.randint(
        50,
        500,
        num_rows
    )
}
```

```python
restaurant_df = pd.DataFrame(restaurant_data)

print(restaurant_df.head())
```

```python
print("Dataset Shape:", restaurant_df.shape)
```

```python
restaurant_df.to_csv(
    'restaurant_sales_5000.csv',
    index=False
)

print("Dataset Saved Successfully")
```
